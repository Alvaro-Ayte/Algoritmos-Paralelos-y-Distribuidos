```python
import random
def generar_conjunto():
    return set(random.sample(range(1, 100), random.randint(8, 15)))
def union(conjuntoA, conjuntoB):
    return list(conjuntoA) + [e for e in conjuntoB if e not in conjuntoA]
def interseccion(conjuntoA, conjuntoB):
    return [e for e in conjuntoA if e in conjuntoB]
    return conjuntoA & conjuntoB
def diferencia_simetrica(conjuntoA, conjuntoB):
    return [e for e in conjuntoA if e not in conjuntoB] + [e for e in conjuntoA if e not in conjuntoB]
    return conjuntoA ^ conjuntoB
def producto_cartesiano(conjuntoA, conjuntoB):
    return [(x, y) for x in conjuntoA for y in conjuntoB]
    return {(a, b) for a in conjuntoA for b in conjuntoB}
def elementos_no_intersectados(conjuntoA, conjuntoB):
    con1 = [e for e in conjuntoA if e not in conjuntoB]
    con2 = [e for e in conjuntoA if e not in conjuntoB]
    return con1, con2
    return (conjuntoA - conjuntoB) | (conjuntoB - conjuntoA)
def inclusion(conjuntoA, conjuntoB):
    inc1 = len([e for e in conjuntoA if e not in conjuntoB]) == 0
    inc2 = len([e for e in conjuntoA if e not in conjuntoB]) == 0
    return inc1, inc2
    return conjuntoA <= conjuntoB
def menu():
    print("\nOperaciones con Conjuntos")
    print("1. Unión")
    print("2. Intersección")
    print("3. Diferencia Simétrica")
    print("4. Producto Cartesiano")
    print("5. Elementos que no se intersectan")
    print("6. Inclusión de conjuntos")
    print("7. Generar nuevos conjuntos")
    print("8. Salir")

# Función principal
def main():
    conjuntoA = generar_conjunto()
    conjuntoB = generar_conjunto()

    print(f"Conjunto A: {conjuntoA}")
    print(f"Conjunto B: {conjuntoB}")

    while True:
        menu()
        opcion = input("\nSelecciona una opción 1 a 8--> ")
        if opcion == '1':
            print('Unión:', set(union(conjuntoA, conjuntoB)))
        elif opcion == '2':
            print('Intersección:', set(interseccion(conjuntoA, conjuntoB)))
        elif opcion == '3':
            print('Diferencia Simétrica:', set(diferencia_simetrica(conjuntoA, conjuntoB)))
        elif opcion == '4':
            print('Producto Cartesiano:', set(producto_cartesiano(conjuntoA, conjuntoB)))
        elif opcion == '5':
            no_intersectan = elementos_no_intersectados(conjuntoA, conjuntoB)
            print("Elementos en conjunto 1 que no están en conjunto 2:", set(no_intersectan[0]))
            print("Elementos en conjunto 2 que no están en conjunto 1:", set(no_intersectan[1]))
        elif opcion == '6':
            incl1, incl2 = inclusion(conjuntoA, conjuntoB)
            print(f"Conjunto 1 está incluido en conjunto 2?: {incl1}")
            print(f"Conjunto 2 está incluido en conjunto 1?: {incl2}")
        elif opcion == '7':
            conjuntoA = generar_conjunto()
            conjuntoB = generar_conjunto()
            print("Nuevos conjuntos generados.")
        elif opcion == 8:
          print('Saliendo...')
        else:
          print('Opción no válida')
```
