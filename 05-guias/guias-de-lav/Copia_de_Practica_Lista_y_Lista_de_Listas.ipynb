**Problema 1**.- Se tiene la relación de los integrantes del equipo de Fútbol de IIS, También se tiene la relación de integrantes del equipo de Basketball.
Escribir una aplicación que determine la relacion de todos los deportistas; otra relacion con los deportistas que participan en ambos equipos.

```python
# --- Modulos de la aplicacion
# Definición de la función para ingresar los equipos
def equipo():
    # Cantidad de miembros para fútbol
    cantidadFutbol = int(input("Ingrese la cantidad de participantes de fútbol--> "))

    futbol = []
    for k in range(cantidadFutbol):
        nombre = input(f"Nombre del participante de fútbol {k+1}: ")
        futbol.append(nombre)

    # Cantidad de miembros para baloncesto
    cantidadBasquet = int(input("Ingrese la cantidad de participantes de basket-->"))

    basket = []
    for k in range(cantidadBasquet):
        nombre = input(f"Nombre del participante de basket {k+1}: ")
        basket.append(nombre)

    # Encontrar los participantes que juegan tanto fútbol como baloncesto
    ambos = list(set(futbol) and set(basket))
    return futbol, basket, ambos

# Función para mostrar lista de integrantes de fútbol
def ListaFutbol(futbol):
    print("Deportistas que son de fútbol:")
    for x in futbol:
        print(x)

# Función para mostrar lista de integrantes de baloncesto
def ListaBasquet(basquet):
    print("Deportistas que son de baloncesto:")
    for x in basket:
        print(x)

# Función para mostrar lista de deportistas que practican ambos deportes
def ListaAmbos(ambos):
    print("Deportistas que practican ambos deportes:")
    for x in ambos:
        print(x)

# Función para mostrar el menú de opciones
def Menu():
    print()
    print('*****   Deportistas   *****')
    print('1. Lista de integrantes de fútbol')
    print('2. Lista de integrantes de basket')
    print('3. Lista de deportistas que practican ambos deportes')
    print('4. FIN')
    opcion = input("Digite una opción: ")
    return opcion
```

```python
+# Programa principal
futbol, basket, ambos = equipo()

while True:
    opcion = Menu()

    if opcion == '1':
        ListaFutbol(futbol)
    elif opcion == '2':
        ListaBasquet(basket)
    elif opcion == '3':
        ListaAmbos(ambos)
    elif opcion == '4':
        print("Gracias por usar el programa.")
        break
    else:
        print("Opción inválida. Por favor, elija una opción válida (1-4).")
```

**Problema 2**.- Implementar una aplicación que permita agregar libros a una lista con la sgte estructura:
CodigoLibro, Titulo, Autor, Especialida, Año
Luego que permita gestionar la información mediante un Menú.

```python
# ************   MÓDULOS ESPECÍFICOS DE LA APLICACIÓN  ***************

# -----------  Menu  -------------
def Menu():
    # -- Mostrar el menú de opciones
    print()
    print('*****   ALUMNOS   *****')
    print('1 Agregar Libros')
    print('2 Consultar Libro')
    print('3 Eliminar Libro')
    print('4 Listar Libros')
    print('5 Listar Libros Por Especialidad')
    print('6 Listar Libros Por Año')
    print('7 FIN')
```

```python
# -----------  Módulo Posicion  -------------
def Posicion(Codigo, Lista):
    P = -1
    I = 0
    while P == -1 and (I < len(Lista)):
        if (Lista[I][0] == Codigo):
            P = I
        else:
            I += 1
    return P
```

```python
# -----------  Módulo Agregar  -------------
def Agregar(Lista):
    Codigo = input('Ingresar el codigo del libro --->')
    if Posicion(Codigo, Lista) >= 0:
        print('El código ya existe en la lista')
    else:
        Titulo = input('Ingresar el titulo del libro: ')
        Autor = input('Ingresar el autor del libro: ')
        Año = int(input('Ingresar el año de publicacion del libro: '))
        Especialidad = input('Ingresar la especialidad del libro: ')
        Lista.append([Codigo, Titulo, Autor, Año, Especialidad])
```

```python
# -----------   Módulo Consultar  -------------
def Consultar(Lista):
    Codigo = input('Ingresar el codigo del libro --->')
    P = Posicion(Codigo, Lista)
    if P >= 0:
        print('Codigo \t Titulo \t Autor \t Año \t Especialidad ')
        print(Lista[P])
    else:
        print(Codigo, 'El codigo del libro no existe')
```

```python
# -----------  Módulo Eliminar  -------------
def Eliminar(Lista):
    Codigo = input('Ingresar el código del libro: ')
    P = Posicion(Codigo, Lista)
    if P >= 0:
        del Lista[P]
    else:
        print(Codigo, 'El codigo del libro no existe')
```

```python
# -----------  Módulo Listar Libros  -------------
def Lista_general(Lista):
    print('Listado general de los libros')
    print('=============================')
    print('Codigo \t Titulo \t Autor \t Año \t Especialidad ')
    for a in Lista:
        print(a[0], '\t', a[1], '\t', a[2], '\t', a[3], '\t', a[4], '\t')
```

```python
# -----------  Módulo Listar Libros Por Especialidad  -------------
def Listar_Especialidad(Lista):
    Especialidad = input('Ingresar la especialidad a enlistar: ')
    encontrado = False
    print('LISTA DE ESPECIALIDAD DE LIBROS')
    print('===============================')
    print('Codigo \t Titulo \t Autor \t Año \t Especialidad ')
    for a in Lista:
        if a[4] == Especialidad:
            print(a[0], '\t', a[1], '\t', a[2], '\t', a[3], '\t', a[4], '\t')
            encontrado = True
    if not encontrado:
        print('La especialidad no existe en la lista')
```

```python
# -----------  Módulo Listar Libros Por Año -------------
def Listar_Año(Lista):
    Año = int(input('Ingresar el año a enlistar: '))
    encontrado = False
    print('LISTA DE AÑOS DE LIBROS')
    print('===============================')
    print('Codigo \t Titulo \t Autor \t Año \t Especialidad ')
    for a in Lista:
        if a[3] == Año:
            print(a[0], '\t', a[1], '\t', a[2], '\t', a[3], '\t', a[4], '\t')
            encontrado = True
    if not encontrado:
        print('No hay libros publicados en el año especificado')
```
